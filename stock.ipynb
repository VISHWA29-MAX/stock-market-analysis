```python
import matplotlib.pyplot as plt
```

```python
stock["Market Category"].value_counts().plot(kind="bar")

plt.title("Market Category Distribution")
plt.xlabel("Market Category")
plt.ylabel("Count")

plt.show()
```

```python
stock["Financial Status"].value_counts().plot(kind="bar")

plt.title("Financial Status Distribution")
plt.xlabel("Financial Status")
plt.ylabel("Count")

plt.show()
```

```python
plt.hist(stock["Round Lot Size"], bins=10)

plt.title("Round Lot Size Distribution")
plt.xlabel("Round Lot Size")
plt.ylabel("Frequency")

plt.show()
```

```python
stock["Market Category"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Market Category Distribution")
plt.ylabel("")

plt.show()
```
